This notebook is a warm-up exercise before we get into actual machine learning models.
The goal is to practice the basic descriptive statistics that show up everywhere in
data analysis and ML: the sample **mean**, **median**, **variance**, and **standard
deviation**. We also practice a very common first step in any data project — cleaning
a dataset by removing rows we don't want to include in the analysis.

## Task

We are given a file with the average monthly salary and population of Russian regions
as of January 1, 2019 (source: Russian Federal State Statistics Service). The original
file can be downloaded here:
[salary_and_population.csv](https://drive.google.com/file/d/1YrEIn6JNWW0RG0r8nV0P6S55TBXcxSna/view?usp=drive_link).

Two regions — **Vologda Region** and **Murmansk Region** — were included in the file by
mistake and must be excluded before we compute any statistics.

Using the cleaned sample, we need to find:

1. The sample **mean** of the salary
2. The sample **median** of the salary
3. The **variance** estimate $S^2$ of the salary
4. The **standard deviation** $\sigma$ of the salary


## 0. Setup

Only two libraries are needed for this exercise:

- **numpy** — for numerical helper functions (not strictly required here since pandas
  already exposes `.mean()`, `.median()`, etc., but it's good practice to have it
  available for any array-level math).
- **pandas** — for loading and manipulating the tabular data.


```python
import numpy as np
import pandas as pd
```

## 1. Load the data

> 2. Place it in a `data/` folder next to this notebook (or update `DATA_PATH` below).


```python
DATA_PATH = "data/salary_and_population.csv"

data = pd.read_csv(DATA_PATH, encoding="utf-8", delimiter=",")

# The source file contains one column with no header (an artifact of how it was
# exported), which pandas reads in as "Unnamed: 4". It carries no information, so
# we drop it right away.
data.drop(columns=["Unnamed: 4"], inplace=True)

data.head()
```

## 2. Inspect the region names

Before filtering, let's check the full list of unique region names in the (Russian)
`Region_RU` column so we know exactly how "Vologda Region" and "Murmansk Region" are
spelled in the source data.


```python
data["Region_RU"].unique()
```

## 3. Remove the two excluded regions

The task states that Vologda Region (`Вологодская область`) and Murmansk Region
(`Мурманская обл`) were mistakenly included and must be dropped before we compute any
statistics.


```python
data = data.loc[
    (data["Region_RU"] != "Вологодская область")
    & (data["Region_RU"] != "Мурманская обл")
]

# Sanity check: the sample size should have shrunk by exactly two rows.
data.shape
```

## 4. Compute the descriptive statistics

With the cleaned sample in hand, we can compute the four requested statistics directly
with pandas' built-in aggregation methods. Note that `.var()` and `.std()` use
`ddof=0` (the *population* formula, dividing by $n$) rather than the default `ddof=1`
(the *sample* formula, dividing by $n-1$), matching how the exercise defines the
variance estimate $S^2$.


```python
mean_salary = round(data["AVG_Salary"].mean(), 2)
print(f"Sample mean of the salary: {mean_salary}")
```

```python
median_salary = round(data["AVG_Salary"].median(), 2)
print(f"Sample median of the salary: {median_salary}")
```

```python
variance_salary = round(data["AVG_Salary"].var(ddof=0), 2)
print(f"Variance estimate of the salary (S^2): {variance_salary}")
```

```python
std_salary = round(data["AVG_Salary"].std(ddof=0), 2)
print(f"Standard deviation of the salary (sigma): {std_salary}")
```

## Summary

- After removing Vologda Region and Murmansk Region, the sample mean salary (~38,658)
  is noticeably higher than the median (~32,473). This gap between mean and median is a
  classic sign of a **right-skewed distribution** — a small number of regions (e.g.
  Moscow, oil/gas regions) have unusually high average salaries and pull the mean
  upward, while most regions cluster at lower values.
- The large variance / standard deviation (~16,447) confirms there is substantial
  spread in regional salaries across Russia.
- Takeaway for later exercises: whenever mean and median disagree by a lot, it's worth
  plotting the distribution (e.g. a histogram) before trusting the mean as a "typical"
  value.
